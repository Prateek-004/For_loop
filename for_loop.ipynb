**Q.(9)** When interest compounds **q** times per year at an annual rate of **r %** for **n years**, the **principle p** compounds to an amount a as per the following fomulas:

*   a = p(1+r/(100*q))**(n*q)

Write a progam to read 10 sets of p,q,r & n and calculate the corresponding as.

**Q.(10)** Write a program to generate all pythagorean Triplets with side length less than or equal to 30 .

```python
for a in range(1,31):
  for b in range(1,31):
    c=(a**2+b**2)**0.5
    if(c>30):
      break
    else:
      print("(",a,",",b,",",c,")")
```

**Q.(13)** Write a program to print 24 hours of day with suitable suffixes like AM, PM, Noon and Midnight.

```python
for hour in range(1,24):
  for minute in range(00,60):

    if hour in range(1,12):        # For day time
      if hour in range(5,11):
        print(hour,":", minute,"AM","Morning")
      elif hour in range(11,12):
        print(hour, ":", minute, "AM", "Noon")
      else:
        print(hour,":", minute, "AM","Night")

    else:                          # For night time
      if hour in range(12,17):
        print(hour,":", minute,"PM","Noon")
      elif hour in range (17,22):
        print(hour, ":", minute, "PM", "Evening")
      else:
        print(hour,":", minute, "PM","Night")
```
